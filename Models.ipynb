```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pathlib
import keras

shared_notebook_dir = "G:/Shared drives/CS229 Project"
dir = pathlib.Path(shared_notebook_dir) / "Code" / "For_Model_Training"
```

```python
# import data from Code\For_Model_Training
X_train = pd.read_pickle(dir / "X_train.pkl")
y_train = pd.read_pickle(dir / "y_train.pkl")
```

```python
X_validation = pd.read_pickle(dir / "X_validation.pkl")
y_validation = pd.read_pickle(dir / "y_validation.pkl")
X_test = pd.read_pickle(dir / "X_test.pkl")
y_test = pd.read_pickle(dir / "y_test.pkl")
```

```python
x_unlabelled = pd.read_pickle(dir / "X_unlabelled.pkl")
```

```python
# Train a simple neural network model sequentially: Baseline with 2 hidden layers having 10 units each
model = keras.Sequential([
    keras.layers.Dense(10, activation='relu', input_shape=(X_train.shape[1],)),
    keras.layers.Dense(10, activation='relu'),
    keras.layers.Dense(1, activation = 'sigmoid')  # Linear activation for regression
])
model.compile(optimizer='adam',
              loss='mse',  # Mean squared error for regression
              metrics=['mae'])  # Mean absolute error as metric
```

```python
# Check data shapes and types before training
print("Data shapes:")
print(f"  X_train: {X_train.shape}")
print(f"  y_train: {y_train.shape}")
print(f"  X_validation: {X_validation.shape}")
print(f"  y_validation: {y_validation.shape}")

print("\nData types:")
print(f"  X_train dtypes:\n{X_train.dtypes}")
print(f"  y_train dtype: {y_train.dtype if hasattr(y_train, 'dtype') else type(y_train)}")

print("\nModel summary:")
model.summary()

print("\nSample X_train values:")
print(X_train.head())
print("\nSample y_train values:")
print(y_train.head())
```

```python
# Convert DataFrames to NumPy arrays with proper dtype
X_train = X_train.astype('float32').values
y_train = y_train.astype('float32').values
X_validation = X_validation.astype('float32').values
y_validation = y_validation.astype('float32').values
X_test = X_test.astype('float32').values
y_test = y_test.astype('float32').values

print("Data converted to NumPy arrays:")
print(f"  X_train: {X_train.shape}, dtype: {X_train.dtype}")
print(f"  y_train: {y_train.shape}, dtype: {y_train.dtype}")
print(f"  X_validation: {X_validation.shape}, dtype: {X_validation.dtype}")
print(f"  y_validation: {y_validation.shape}, dtype: {y_validation.dtype}")
print(f"  X_test: {X_test.shape}, dtype: {X_test.dtype}")
print(f"  y_test: {y_test.shape}, dtype: {y_test.dtype}")
```

```python
#train the model use X_validation and y_validation for validation
model.fit(X_train, y_train, epochs=50, batch_size=32, validation_data=(X_validation, y_validation))
```

```python
#save the model as baseline_model.keras
model.save(dir / "baseline_model.keras")
#load the model
model = keras.models.load_model(dir / "baseline_model.keras")
```

```python
# Calculate MSE for all three datasets
predictions_test = model.predict(X_test)
mse_test = np.mean((predictions_test.flatten() - y_test) ** 2)

predictions_validation = model.predict(X_validation)
mse_validation = np.mean((predictions_validation.flatten() - y_validation) ** 2)

predictions_train = model.predict(X_train)
mse_train = np.mean((predictions_train.flatten() - y_train) ** 2)

print(f"Mean Squared Error on test set: {mse_test:.6f}")
print(f"Mean Squared Error on validation set: {mse_validation:.6f}")
print(f"Mean Squared Error on training set: {mse_train:.6f}")
```

```python
# Analyze bias-variance tradeoff
print("Bias-Variance Analysis:")
print("="*50)

if mse_train < mse_validation * 0.7:  # Training MSE much lower than validation
    print("⚠️ HIGH VARIANCE (Overfitting) detected!")
    print("   Training MSE is significantly lower than validation/test MSE")
    print("\nRecommended solutions:")
    print("   1. Add dropout layers")
    print("   2. Add L2 regularization (weight decay)")
    print("   3. Reduce model complexity (fewer/smaller layers)")
    print("   4. Get more training data")
    print("   5. Add batch normalization")
elif mse_train > mse_validation * 1.3:  # Training MSE much higher than validation
    print("⚠️ HIGH BIAS (Underfitting) detected!")
    print("   Training MSE is high and similar to validation MSE")
    print("\nRecommended solutions:")
    print("   1. Increase model complexity (more/larger layers)")
    print("   2. Train for more epochs")
    print("   3. Add more features")
    print("   4. Reduce regularization")
else:
    print("✓ Model appears to have good bias-variance balance")

print(f"\nMSE Ratios:")
print(f"   Validation/Train: {mse_validation/mse_train:.3f}")
print(f"   Test/Train: {mse_test/mse_train:.3f}")
```

```python
# Tune L2 regularization parameter (lambda) based on validation loss
lambda_values = [0.0, 0.001, 0.005, 0.01, 0.05, 0.1]
results = []

print("Tuning L2 Regularization (lambda):")
print("="*60)

for lambda_val in lambda_values:
    # Build model with current lambda
    model_reg = keras.Sequential([
        keras.layers.Dense(10, activation='relu', input_shape=(X_train.shape[1],), 
                          kernel_regularizer=keras.regularizers.l2(lambda_val)),
        keras.layers.BatchNormalization(),
        
        keras.layers.Dense(10, activation='relu',
                          kernel_regularizer=keras.regularizers.l2(lambda_val)),
        keras.layers.BatchNormalization(),
        
        keras.layers.Dense(1, activation='sigmoid')  # sigmoid activation for regression
    ])
    
    model_reg.compile(optimizer='adam', loss='mse', metrics=['mae'])
    
    # Train the model
    history = model_reg.fit(
        X_train, y_train, 
        epochs=50, 
        batch_size=32, 
        validation_data=(X_validation, y_validation),
        verbose=0
    )
    
    # Evaluate on all datasets
    pred_train = model_reg.predict(X_train, verbose=0)
    mse_train_lambda = np.mean((pred_train.flatten() - y_train) ** 2)
    
    pred_val = model_reg.predict(X_validation, verbose=0)
    mse_val_lambda = np.mean((pred_val.flatten() - y_validation) ** 2)
    
    pred_test = model_reg.predict(X_test, verbose=0)
    mse_test_lambda = np.mean((pred_test.flatten() - y_test) ** 2)
    
    results.append({
        'lambda': lambda_val,
        'train_mse': mse_train_lambda,
        'val_mse': mse_val_lambda,
        'test_mse': mse_test_lambda,
        'model': model_reg
    })
    
    print(f"λ={lambda_val:6.3f} | Train: {mse_train_lambda:.6f} | Val: {mse_val_lambda:.6f} | Test: {mse_test_lambda:.6f}")

# Find best lambda based on validation MSE
best_result = min(results, key=lambda x: x['val_mse'])
best_lambda = best_result['lambda']
model_regularized = best_result['model']

print("\n" + "="*60)
print(f"Best λ = {best_lambda} (lowest validation MSE)")
print(f"\nBest Regularized Model Performance:")
print(f"  Training MSE:   {best_result['train_mse']:.6f}")
print(f"  Validation MSE: {best_result['val_mse']:.6f}")
print(f"  Test MSE:       {best_result['test_mse']:.6f}")

print("\nComparison with baseline:")
print(f"  Training improvement:   {((mse_train - best_result['train_mse'])/mse_train*100):+.2f}%")
print(f"  Validation improvement: {((mse_validation - best_result['val_mse'])/mse_validation*100):+.2f}%")
print(f"  Test improvement:       {((mse_test - best_result['test_mse'])/mse_test*100):+.2f}%")

# Store best model metrics for later use
mse_train_reg = best_result['train_mse']
mse_validation_reg = best_result['val_mse']
mse_test_reg = best_result['test_mse']
```

```python
#save the best model as regularized_model.keras
model_regularized.save(dir / "regularized_model.keras")
```
